```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
df= pd.read_csv("thyroid_dataset.csv")
```

```python
df.head()
```

```python
x= df.drop("Outlier_label", axis=1)
y= df["Outlier_label"]
```

```python
scaler= StandardScaler()
x_scaled= scaler.fit_transform(x)
```

```python
from sklearn.ensemble import IsolationForest

clf= IsolationForest(n_estimators= 200, contamination= "auto", random_state= 42)
#if contamination=0.036 then it detect actual outlier bcoz in this data set 0.036% outliers
```

```python
labels= clf.fit_predict(x_scaled)
```

```python
labels
```

```python
# Visualize
from sklearn.decomposition import PCA

pca= PCA(n_components=2)
x_pca= pca.fit_transform(x_scaled)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(x_pca[:, 0], x_pca[:, 1], c=labels)
plt.xlabel("Pc1")
plt.ylabel("PC2")
```

```python
import numpy as np

n_outliers= np.sum(labels== -1)
n_normal= np.sum(labels== 1)

print("oytlier= ", n_outliers)
print("normal= ", n_normal)
```

# LOF: Local Outlier Factor

```python
from sklearn.neighbors import LocalOutlierFactor
```

```python
neighbs= LocalOutlierFactor(contamination= 0.036)

labels= neighbs.fit_predict(x_scaled)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(x_pca[:, 0], x_pca[:, 1], c=labels)
plt.xlabel("Pc1")
plt.ylabel("PC2")
```

```python
n_outliers= np.sum(labels== -1)
n_normal= np.sum(labels== 1)

print("oytlier= ", n_outliers)
print("normal= ", n_normal)
```
